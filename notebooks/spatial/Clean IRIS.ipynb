```python
import geopandas as gpd
import pandas as pd
import py7zr, tempfile
```

```python
iris_path = "../../resources/iris/CONTOURS-IRIS_2-1__SHP__FRA_2017-06-30.7z"

departments = ["01", "38", "42", "69"]

output_path = "../../results/spatial/iris.parquet"
```

```python
with tempfile.TemporaryDirectory() as directory:
    with py7zr.SevenZipFile(iris_path) as archive:
        archive.extractall(directory)

        df_iris = gpd.read_file("/".join([
            directory, 
            "CONTOURS-IRIS_2-1__SHP__FRA_2017-06-30",
            "CONTOURS-IRIS",
            "1_DONNEES_LIVRAISON_2016",
            "CONTOURS-IRIS_2-1_SHP_LAMB93_FE-2016",
            "CONTOURS-IRIS.shp"
        ]))
```

```python
# Filter departments
df_iris = df_iris[df_iris["INSEE_COM"].str[:2].isin(departments)]
```

```python
df_iris = df_iris.rename(columns = {
    "CODE_IRIS": "iris"
})[["iris", "geometry"]]
```

```python
df_iris["iris"] = df_iris["iris"].astype(str)
```

```python
df_iris.to_parquet(output_path)
```
